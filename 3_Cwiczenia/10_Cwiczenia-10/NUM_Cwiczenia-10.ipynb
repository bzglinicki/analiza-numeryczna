$$

$$
Jej objętość możemy wyznaczyć numerycznie na podstawie wzoru
$$
V_n = \int \int \cdots \int _{B_n} \, \mathrm{d}x_1 \mathrm{d}x_2 \cdots \mathrm{d}x_n
= \int \int \cdots \int _{\Omega _n} \chi _n \left( x_1 , \, x_2 , \, \ldots , \, x_n \right) \, \mathrm{d}x_1 \mathrm{d}x_2 \cdots \mathrm{d}x_n ,
$$
gdzie $\Omega _n = \left[ 0 , \, 1 \right] ^n$ oraz
$$
\chi _n \left( x_1 , \, x_2 , \, \ldots , \, x_n \right) = \begin{cases}
1 , & \text{gdy } \left( x_1 , \, x_2 , \, \ldots , \, x_n \right) \in B_n , \\
0 & \text{w przeciwnym przypadku} .
\end{cases}
$$

$$
