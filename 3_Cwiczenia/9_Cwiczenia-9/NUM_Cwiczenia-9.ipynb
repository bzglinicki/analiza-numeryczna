Niech $S^T_N$ i $S^S_N$ będą aproksymacjami wartości tej całki uzyskanymi za pomocą, odpowiednio, złożonej kwadratury trapezów i złożonej kwadratury Simpsona, opartych na podziale przedziału $\left[ a , \, b \right]$ na $N$ podprzedziałów tej samej długości $h = \left( b - a \right) / N$.

Błąd przybliżenia wartości całki $I$ uzyskanego za pomocą złożonej kwadratury Simpsona, $S^S_N$, jest mniej więcej $h^{-2}$ razy mniejszy od błędu przybliżenia wartości tej całki za pomocą złożonej kwadratury trapezów, $S^T_N$. Sugeruje to, że wielkość
$$
E_N = \left| S^S_N - S^T_N \right|
