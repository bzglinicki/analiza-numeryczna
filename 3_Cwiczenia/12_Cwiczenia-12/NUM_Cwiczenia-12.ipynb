## Zagadnienie brzegowe: metoda strzałów (*Shooting Method*)

Idea metody strzałów polega na przekształceniu zagadnienia brzegowego w zagadnienie początkowe.

Załóżmy, że mamy warunki brzegowe w postaci
$$
y(a) = y_a , \qquad y(b) = y_b . 
$$
Staramy się wówczas tak dobrać wartość $y'_a$, aby rozwiązanie tego samego równania różniczkowego z warunkami początkowymi
$$
y(a) = y_a , \qquad y'(a) = y'_a . 
$$
spełniało warunek $y(b) = y_b$.

**Algorytm.**
1. Zdefiniuj funkcję $F(s)$, która każdej wartości $s = y'(a)$ przypisuje błąd $y(b) - y_b$.
2. Wyznacz $s_0$ takie, że $F(_0) = 0$.
3. Rozwiąż zagadnienie początkowe z $y'(a) = s_0$ — będzie to rozwiązanie zagadnienia brzegowego.

## Zagadnienie brzegowe: metoda różnic skończonych (*Finite Difference Method*)

Metoda różnic skończonych polega na dyskretyzacji dziedziny $[a, b]$ na $n$ węzłów i zastąpieniu pochodnych w równaniu różniczkowym ich różnicowymi przybliżeniami:

\begin{align*}
y'(x_i) \, & \approx \, \frac{y_{i+1} - y_{i-1}}{2\Delta x} , \\[5px]

y''(x_i) \, & \approx \, \frac{y_{i-1} - 2y_i + y_{i+1}}{\Delta x^2} .
\end{align*}

Po podstawieniu do równania różniczkowego, dla każdego węzła wewnętrznego otrzymujemy jedno równanie liniowe. Razem z warunkami brzegowymi tworzą układ $A\mathbf{y} = \mathbf{b}$, który rozwiązujemy metodami algebry liniowej.

### Zadanie 4. Warunki Dirichleta, Neumanna i Robina.

W metodzie różnic skończonych warunek brzegowy "wbudowujemy" do macierzy $A$:

- **Dirichlet** ($y(a) = c$): bezpośrednio $A[0,0]=1$, $b[0]=c$
- **Neumann** ($y'(a) = c$): różnica progresywna: $-y_1 + y_2 = c\,\Delta x$
- **Robin** ($\alpha y(a) + \beta y'(a) = g$): kombinacja liniowa

Metoda "ghost nodes" pozwala uzyskać aproksymację drugiego rzędu dla warunków Neumanna, traktując punkt wirtualny $y_0$ poza dziedziną.

### Zadanie 7. Nieliniowe zagadnienie brzegowe: iteracja sukcesywna.

Gdy zagadnienie brzegowe jest nieliniowe, macierz $A$ nie może być stała — pojawia się zależność od $y_i$. Rozwiązanie przebiega iteracyjnie:

1. Rozwiąż zagadnienie brzegowe bez członu nieliniowego → otrzymaj $y^{(0)}$.
2. Używając $y^{(k)}$ jako stałej w członie nieliniowym, zbuduj i rozwiąż układ → $y^{(k+1)}$.
3. Powtarzaj aż do zbieżności: $\max|y^{(k+1)} - y^{(k)}| < \varepsilon$.

Rozwiąż nieliniowe zagadnienie brzegowe

$$y'' - 3y - x^2 - 100y^2 = 0, \quad y(0) = 0, \quad y(1) = 0.$$
